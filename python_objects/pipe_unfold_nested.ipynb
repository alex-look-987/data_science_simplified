# Unfold Nested Iterables with Pipe

```python
nested = [[1, 2, [3]], [4, 5]]
```

## Chain

Chain a Sequence of Iterables

```python
from pipe import chain 
```

```python
list(nested | chain)
```

## Traverse

```python
from pipe import traverse
```

```python

list(nested | traverse)
```

## Traverse Integration with Select

```python
from pipe import select
```

```python
fruits = [
    {'name': 'apple', 'price': [2, 5]},
    {'name': 'orange', 'price': 4},
    {'name': 'grape', 'price': 5}
]
```

```python
list(
    fruits
    | select(lambda fruit: fruit['price'])
    | traverse
)
```
