# collection.Counter: Count the Occurrences of Items in a List

Counting the occurrrencess of each item in a list using a for-loop is slow and inefficient

```python
char_list = ['a', 'b', 'c', 'a', 'd', 'b', 'b']
```

```python
def custom_counter(char_list: list):
    char_counter = {}

    for char in char_list:
        if char not in char_counter:
            char_counter[char] = 1
        else:
            char_counter[char] +=1

    return char_counter
```

```python
from collections import Counter
```

```python
Counter(char_list)
```

Compare the Execution Time Between Two Approaches

```python
import random
from timeit import timeit

random.seed(0)
```

```python
num_list = [random.randint(0, 22) for _ in range(1_000)]

numExp = 100
```

```python
custom_time = timeit("custom_counter(num_list)", globals=globals())
counter_time = timeit("Counter(num_list)", globals=globals())

print(custom_time / counter_time)
```
